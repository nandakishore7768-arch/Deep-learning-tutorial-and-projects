```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import make_classification

x,y = make_classification(n_samples = 100, n_features = 2, n_informative=1, n_redundant=1, n_classes=2,n_clusters_per_class=1, random_state=42,class_sep=10, hypercube=False)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train
```

```python
np.random.shuffle(x_train)
x_train
```

```python
class Perceptron:
    def __init__(self, learning_rate = 0.01, epochs = 1000):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None

    def func(self, x):
        return 1 if x > 0 else 0

    def fit(self, x, y):
        x_ = np.insert(x, 0, 1, axis = 1)
        self.weights = np.ones(x_.shape[1])
        for i in range(self.epochs):
            indices = np.random.permutation(x_.shape[0])
            for k in indices:
                xi = x_[k]
                yi = y[k]

                y_pred = self.func(np.dot(xi, self.weights))
                y_diff = yi - y_pred
                xi_ = y_diff * xi
                self.weights = self.weights + self.lr * xi_

        return self

    def predict(self, x):
        x_ = np.insert(x, 0, 1, axis = 1)
        y_pred = np.dot(x_, self.weights) > 0
        return y_pred
         
```

```python
model = Perceptron(learning_rate=0.01, epochs = 1000)
model.fit(x_train, y_train)
```

```python
pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, pred)
```
