```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras import Sequential
from tensorflow.keras.datasets import imdb
from tensorflow.keras.layers import SimpleRNN, Embedding,Dense, Input
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words = 10000)

x_train = pad_sequences(x_train, maxlen = 100)
x_test = pad_sequences(x_test, maxlen = 100)
```

```python
model = Sequential()
model.add(Input(shape = (32, )))
model.add(Embedding(input_dim = 10000,output_dim =  32, input_length = 100))
model.add(SimpleRNN(5, return_sequences=True))
model.add(SimpleRNN(5))
model.add(Dense(1, activation = "sigmoid"))
```

```python
model.summary()
```

```python
from tensorflow.keras.layers import LSTM, GRU
```

```python
model1 = Sequential()
model1.add(Input(shape = (100, )))
model1.add(Embedding(input_dim = 10000,output_dim =  32, input_length = 100))
model1.add(LSTM(32, return_sequences=True))
model1.add(LSTM(32))
model1.add(Dense(1, activation = "sigmoid"))
```

```python
model1.summary()
```

```python
model2 = Sequential()
model2.add(Input(shape = (100, )))
model2.add(Embedding(input_dim = 10000, output_dim = 32, input_length = 100))
model2.add(GRU(5, return_sequences=True))
model2.add(GRU(5))
model2.add(Dense(1, activation = "sigmoid"))
```

```python
model2.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
pred = model.predict(x_test)
```
