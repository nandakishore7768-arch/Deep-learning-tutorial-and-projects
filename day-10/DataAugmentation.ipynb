```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.utils import load_img, img_to_array
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
image =  load_img("/content/cat.webp", target_size = (200, 200))
```

```python
import matplotlib.pyplot as plt

plt.imshow(image)
```

```python
img = img_to_array(image)
```

```python
datagen = ImageDataGenerator(
    rotation_range = 20, # angle
    zoom_range = 0.5, # distortion
    horizontal_flip = True,
    vertical_flip = False,
)
```

```python
img.shape
```

```python
input_img = img.reshape(1, 200, 200, 3)
```

```python
i = 0

for output in datagen.flow(input_img, batch_size = 1, save_to_dir = "/content/output"):
  i += 1

  if(i > 19):
    break
```
